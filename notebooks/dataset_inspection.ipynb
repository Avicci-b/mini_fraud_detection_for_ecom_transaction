# Phase 2 : Understand the data
In this phase we inspect the raw dataset before doing EDA or modeling.

We will check:
- number of rows and columns
- data types
- missing values
- duplicate rows
- summary statistics
- timestamp ranges
- target distribution

```python
import pandas as pd 
import numpy as np
```

```python
# 1. Import the dataset
fraud_df = pd.read_csv("../data/raw/mini_transaction.csv")


fraud_df.shape
print()
fraud_df.head(5)
```

```python
# 2. Columns and data types
print(fraud_df.dtypes)
print()
print("Number of columns: " , len(fraud_df.columns))
print("Columns: ", list(fraud_df.columns))
```

The following cell is important. This is beacuse I am going to drop the `ip_address` column since I am not going to do the merging the Ip_address here with another dataset of Ip_Address ranges. This is because the soul purpose of this project is to examine why and how the models work and then how SMOTE and SHAP are done. I am going to use the other features for the modeling part.

```python
# Dropping the ip_address column from the fraud_df dataset
fraud_df.drop("ip_address", axis=1, inplace=True)
```

```python
# check if the ip_address is dropped
fraud_df.columns
```

```python
# -----------------------------
# 4) Missing values
# -----------------------------
missing = fraud_df.isnull().sum()
print("Missing values per column:")
print(missing[missing > 0])
print()
print("Total missing values:", fraud_df.isnull().sum().sum())
```

```python
# -----------------------------
# 5) Duplicate rows
# -----------------------------
duplicate_count = fraud_df.duplicated().sum()
print("Duplicate rows:", duplicate_count)

if duplicate_count > 0:
    print(fraud_df[fraud_df.duplicated(keep=False)].head())
```

```python
# -----------------------------
# 6) Summary statistics
# -----------------------------
numeric_cols = fraud_df.select_dtypes(include=[np.number]).columns
print("Numerical columns:")
print(fraud_df[numeric_cols].describe().T)
print()

# Optional: show non-numeric columns
print("Non-numeric columns:")
print(fraud_df.select_dtypes(exclude=[np.number]).head())
```

```python
# -----------------------------
# 7) Check timestamp columns
# -----------------------------
# Convert timestamps if needed
for col in ["signup_time", "purchase_time"]:
    if col in fraud_df.columns:
        fraud_df[col] = pd.to_datetime(fraud_df[col])
        print(f"{col} min:", fraud_df[col].min())
        print(f"{col} max:", fraud_df[col].max())
        print()
```

```python
# Verify the type
# One thing that I have learnt here is that with out the print statment here its
# output is in wierd way but the same way as datetime. Try is another time
print(fraud_df["signup_time"].dtype)
print(fraud_df["purchase_time"].dtype)
```

```python
# -----------------------------
# 8) Target distribution
# -----------------------------
if "class" in fraud_df.columns:
    print(fraud_df["class"].value_counts())
    print()
    print(fraud_df["class"].value_counts(normalize=True))
```

```python
# 9. Save the updated dataset to ../data/processed/

fraud_df.to_csv("../data/processed/final_mini_transaction.csv", index=False) 
print("🗂 Final output of the file is done 📂")
```
